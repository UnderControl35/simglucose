## Analysis of data type and length

```python
def analysis_data(data):
    print('Data Type:', type(data))
    if isinstance(data, list): print('Data Length:', len(data))
    if isinstance(data, dict): print('Data Keys:', data.keys())
    if isinstance(data, np.ndarray): print('Data Shape:', data.shape)
    if isinstance(data, tuple): print('Data Length:', len(data))
    if isinstance(data, pd.DataFrame): print('Data Shape:', data.shape)
    if isinstance(data, pd.Series): print('Data Shape:', data.shape)
    
    # Convert to DataFrame with column names
    df = pd.DataFrame(data, columns=['observations', 'next_observations', 'actions', 'rewards', 'terminals'])
    #df.head()
    return df
```

## Visualize the .csv file 

```python
df = pd.read_csv(output_file)
# Extract the first value from each observation array
# (Assuming they're stored as strings like "[[143.25545]]", we'll strip brackets)
# Extract observations (assuming a single row with a string of arrays)
# Remove brackets and split into individual values
obs_string = df['observations'].iloc[0].replace('[', '').replace(']', '').split()
observations = [float(val) for val in obs_string if val.strip()]

# Extract rewards (space-separated values in a single string)
rewards_string = df['rewards'].iloc[0].strip('[] ').split()
rewards = [float(val) for val in rewards_string if val.strip()]

# Ensure lengths match (trim if necessary)
min_length = min(len(observations), len(rewards))
observations = observations[:min_length]
rewards = rewards[:min_length]

# Create the plot
fig, ax1 = plt.subplots(figsize=(10, 6))

# Plot observations on the left y-axis
ax1.plot(observations, color='blue', label='Observations')
ax1.set_xlabel('Time Step')
ax1.set_ylabel('Observations', color='blue')
ax1.tick_params(axis='y', labelcolor='blue')

# Create a second y-axis for rewards
ax2 = ax1.twinx()
ax2.plot(rewards, color='red', label='Rewards')
ax2.set_ylabel('Rewards', color='red')
ax2.tick_params(axis='y', labelcolor='red')

# Title and grid
plt.title('Observations and Rewards Over Time')
ax1.grid(True, linestyle='--', alpha=0.7)

# Show the plot
plt.tight_layout()
plt.show()
```
